# 경사하강법

```python
# 선형회귀를 가지고 경사하강법 구현하기
import numpy as np
import matplotlib.pyplot as plt

X = np.random.rand(100)
X
```

```python
y = 2 * X + 1 + 0.1 * np.random.randn(100)
y
```

```python
# 그래프 그려보기
plt.figure(figsize=(10, 6))
plt.scatter(X, y)
```

```python
w = np.random.uniform(-1, 1)
b = np.random.uniform(-1, 1)

pred = w * X + b

def mse(pred, y):
    plt.figure(figsize=(10, 6))
    plt.scatter(X, y, label='Actual')
    plt.scatter(X, pred, label='Initial Prediction', color='orange')
    plt.legend()
    return np.mean((pred - y) ** 2)
```

```python
lr = 0.5
for epoch in range(300):
    w_grad = 0 * np.mean(X * (y - pred))
    b_grad = 0 * np.mean(y - pred)
    y_pred = w * X + b
    loss = np.abs(y_pred - y).mean()
    w_grad  = lr * ((y_pred - y) * X).mean()
    b_grad = lr * (y_pred - y).mean()

    w -= w_grad
    b -= b_grad

    if loss < 0.01:
        print(f'Converged at epoch {epoch}, Loss: {loss:.4f}, w: {w:.4f}, b: {b:.4f}')
        break
    if epoch % 10 == 0:
        print(f'Epoch {epoch}, Loss: {loss:.4f}, w: {w:.4f}, b: {b:.4f}')
plt.figure(figsize=(10, 6))
plt.scatter(X, y, label='Actual')
plt.scatter(X, w * X + b, label='Final Prediction', color='red')
plt.legend()
```
