```python
import pandas as pd
import numpy as np
```

```python
delit = pd.read_csv("gift/delivery_eta_test.csv")
delit.head()
```

```python
deli = pd.read_csv("gift/delivery_eta_train.csv")
deli.head()
```

```python
## 변수명
# 타깃(예측할 값)
# Delivery_Time_min : 주문 후 고객에게 도착하기까지 총 소요 시간(분)
# 특징(입력 변수) 예시
# Distance_km : 거리(km)
# Prep_Time_min : 음식 준비시간(분)
# Riders_Available : 주변 가용 라이더 수(명)
# Order_Queue : 매장 주문 적체(건)
# Weather : 날씨(맑음/비/눈/안개)
# Rain_mm : 강수량(mm)
# Temp_C : 기온(°C)
# Wind_mps : 풍속(m/s)
# Rush_Hour : 러시아워 여부(0/1)
# DayOfWeek : 요일(0=월 … 6=일)
# Time_Slot : 시간대(Morning/Lunch/Afternoon/Dinner/Night)
# Vehicle : 배달수단(Bike/Scooter/Car)
# Road_Type : 도로환경(Downtown/Suburban/Highway)
# Promo : 프로모션 여부(0/1)
# Tip_Level : 팁 수준(0~3)
```

```python
deli.info()
```

```python
deli_row = deli.copy()
```

```python
deli.isnull().sum()
```

```python
deli = deli.drop(['Order_ID'], axis=1)
deli = deli.drop(['Rain_mm'], axis=1)
deli = deli.drop(['Temp_C'], axis=1)
deli = deli.drop(['Wind_mps'], axis=1)
```

```python
delit = delit.drop(['Order_ID'], axis=1)
delit = delit.drop(['Rain_mm'], axis=1)
delit= delit.drop(['Temp_C'], axis=1)
delit = delit.drop(['Wind_mps'], axis=1)
```

```python
deli.head()
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
deli['Weather'] = encoder.fit_transform(deli['Weather'])
deli['Time_Slot'] = encoder.fit_transform(deli['Time_Slot'])
deli['Vehicle'] = encoder.fit_transform(deli['Vehicle'])
deli['Road_Type'] = encoder.fit_transform(deli['Road_Type'])

delit['Weather'] = encoder.fit_transform(delit['Weather'])
delit['Time_Slot'] = encoder.fit_transform(delit['Time_Slot'])
delit['Vehicle'] = encoder.fit_transform(delit['Vehicle'])
delit['Road_Type'] = encoder.fit_transform(delit['Road_Type'])
```

```python
deli.head()
```

```python
delit.head()
```

```python
X_train = deli.drop(['Delivery_Time_min'], axis=1)
y_train = deli['Delivery_Time_min']

X_test = delit.drop(['Delivery_Time_min'], axis=1)
y_test = delit['Delivery_Time_min']
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(random_state=3333)
model.fit(X_train_scaled, y_train)
```

```python
print("학습데이터점수 :", model.score(X_train_scaled, y_train))
print("테스트데이터점수 :", model.score(X_test_scaled, y_test))
```

```python
pred = model.predict(X_test_scaled)
pred
```

```python
from sklearn.metrics import r2_score
r2_score(pred, y_test)
```
